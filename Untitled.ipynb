```python
%load_ext wurlitzer
%load_ext autoreload
%autoreload 2

import logging
logging.basicConfig()
logging.getLogger().setLevel(logging.INFO)
    
import os
from ponx.network import Graph
from ponx.simulation import run_matrix_game, RunConfig, EndReason
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
config = RunConfig(max_steps=5000, store_steps=20)
payoffs = [[4.0, 1.0], [3.0, 2.0]]
#payoffs = [[-10.0, -10.0], [10.0, 2.0]]
```

```python
def read_graphs(path):
    names = sorted([name for name in os.listdir(path) if name.endswith(".json")])
    return [Graph.read(os.path.join(path, name)) for name in names]

l_graphs = read_graphs("/home/spirali/projects/netprocess-data/v0.2/data-L")
m_graphs = read_graphs("/home/spirali/projects/netprocess-data/v0.2/data-M")
s_graphs = read_graphs("/home/spirali/projects/netprocess-data/v0.2/data-S")
all_graphs = l_graphs + m_graphs + s_graphs
```

```python
config = RunConfig(max_steps=2500, store_steps=0)
results = run_matrix_game(payoffs, all_graphs, config, repeats=100)
```

```python
set([r.end_reason for r in results if r.end_reason != EndReason.Converged])
```

```python
data = []
for r in results:
    history = r.history
    last = len(history) - 1
    if r.end_reason == EndReason.Converged:
        coop = r.means[-1]
    elif r.end_reason == EndReason.MeanCheck:
        coop = np.mean(r.means[-32:])
    else:
        raise Exception("Invalid endreason")
    name = r.graph.name
    if "data-L" in name:
        gset = "L"
    elif "data-M" in name:
        gset = "M"
    elif "data-S" in name:
        gset = "S"
    else:
        raise Exception("Invalid graph")
    data.append((os.path.basename(r.graph.name), gset, r.graph.metadata["stats"], coop, str(r.end_reason), len(r.means)))

df = pd.DataFrame(data, columns=["graph", "gset", "stats", "coop", "end_reason", "step"])
df = pd.concat([df, pd.json_normalize(df.stats)], axis=1)
df

c = list(df.columns)
c.remove("coop")
c.remove("stats")
c.remove("step")
c.remove("end_reason")

df2 = df.groupby(c)["coop"].mean().reset_index()
df2
```
