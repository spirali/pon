```python
%load_ext wurlitzer
%load_ext autoreload
%autoreload 2

import logging
logging.basicConfig()
logging.getLogger().setLevel(logging.INFO)
    
import os
from ponx.network import Graph
from ponx.simulation import run_matrix_game, RunConfig, EndReason
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
config = RunConfig(max_steps=5000, store_steps=20)
payoffs = [[4.0, 1.0], [3.0, 2.0]]
#payoffs = [[-10.0, -10.0], [10.0, 2.0]]
```

```python
path = "/home/spirali/projects/netprocess-data/v0.2/data-L"
names = sorted([name for name in os.listdir(path) if name.endswith(".json")])
all_graphs = [Graph.read(os.path.join(path, name)) for name in names]
```

```python
config = RunConfig(max_steps=2500, store_steps=1)
results = run_matrix_game(payoffs, all_graphs, config, repeats=10)
```

```python
set([r.end_reason for r in results if r.end_reason != EndReason.Converged])
```

```python
data = []
for r in results:
    history = r.history
    last = len(history) - 1
    if r.end_reason == EndReason.Converged:
        coop = r.means[-1]
    elif r.end_reason == EndReason.MeanCheck:
        coop = np.mean(r.means[-32:])
    else:
        raise Exception("Invalid endreason")
    data.append((os.path.basename(r.graph.name), r.graph.metadata["stats"], coop, str(r.end_reason), len(r.means)))

df = pd.DataFrame(data, columns=["graph", "stats", "coop", "end_reason", "step"])
```
