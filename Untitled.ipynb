```python
%load_ext wurlitzer
%load_ext autoreload
%autoreload 2

import os
from ponx.network import Graph
from ponx.simulation import run_matrix_game, RunConfig
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
config = RunConfig(max_steps=5000, store_steps=1)
payoffs = [[4.0, 1.0], [3.0, 2.0]]
#payoffs = [[-10.0, -10.0], [10.0, 2.0]]
```

```python
path = "/home/spirali/projects/netprocess-data/v0.2/data-L"
names = [name for name in os.listdir(path) if name.endswith(".h5")]
graphs = [Graph.read_h5(os.path.join(path, name)) for name in names]
```

```python
config = RunConfig(max_steps=2500, store_steps=16)
results = run_matrix_game(payoffs, graphs, config, repeats=200)
```

```python
data = []
for r in results:
    data.append((os.path.basename(r.graph.name), np.mean(r.history[-1][1]), r.converged, r.n_steps))

df = pd.DataFrame(data, columns=["graph", "coop", "converged", "n_steps"])
df
```

```python
px.box(df, x="graph", y="n_steps", color="converged")
```
